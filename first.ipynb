```python
import numpy as np # numaral islemler icin kullanilan kütüphane
import pandas as pd # veri isleme ve analiz icin kullanilan kütüphane
import matplotlib.pyplot as plt # veri görselleştirme için kullanılan kütüphane
import seaborn as sns # istatistiksel veri görselleştirme için kullanılan kütüphane
from sklearn.ensemble import RandomForestClassifier # makine ögrenmesi algoritması  
```

```python
df= pd.read_csv("01ilkYZ/glass.csv") # csv dosyasını okuma
```

```python
df.head() # veri setinin ilk 5 satırını gösterme
```

```python
df.shape # veri setinin satır ve sütun sayısını gösterme Zeile und Spaltenzahl des Datensatzes anzeigen
```

```python
df["Type"].value_counts() # "Type" sütunundaki her bir sınıfın kaç kez tekrarlandığını gösterme
```

```python
X = df.drop("Type",axis=1) # "Type" sütununu bağımsız değişkenlerden çıkarma
y = df["Type"] # "Type" sütununu bağımlı değişken olarak
```

```python
from sklearn.model_selection import  train_test_split # veri setini eğitim ve test olarak bölmek için kullanılan fonksiyon
```

```python
X_train, X_test, y_train, y_test= train_test_split(X,y, test_size=0.3,random_state=9) # veri setini %80 eğitim ve %20 test olarak bölme  y test = camin tipi
```

```python
model = RandomForestClassifier() # Random Forest sınıflandırma modelini oluşturma
```

```python
model.fit(X_train, y_train) # MAKINE ÖGRENME ALGORITMASI satiri
```

```python
cam_pred = model.predict(X_test) # test verisi ile tahmin yapma neyi preditct ediyoruz? cam tipini tahmin ediyoruz
```

```python
from sklearn.metrics import ConfusionMatrixDisplay, classification_report,accuracy_score # karışıklık matrisini görselleştirmek için kullanılan sınıf
```

```python
accuracy_score(y_test, cam_pred) # modelin doğruluk oranını hesaplama
```

MODEL geht %85 erfolgreich 

```python
ConfusionMatrixDisplay.from_estimator(model, X_test, y_test) # karışıklık matrisini görselleştirme




```
